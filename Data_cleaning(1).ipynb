```python
import pandas as pd 

data1 = {
    "subject_id": [1, 2, 2, 3, 4, 5, 6, 7, 8, 9],
    "age": [25, 30, 30, None, 22, 999, 28, 31, 27, 24],
    "condition": ["control", "treatment", "treatment", "Control", "treatment", "control", "CONTROL", "treatment", "control", "treatment"],
    "score": [88, 92, 92, 78, 85, 91, 84, 90, 87, None]
}
df_messy1 = pd.DataFrame(data1)
print(df_messy1)
```

```python
df_clean1 = df_messy1.copy()
```

```python
df_clean1 = df_clean1.drop_duplicates()
```

```python
df_clean1["condition"] = df_clean1["condition"].str.lower()
```

```python
df_clean1.loc[df_clean1["age"] == 999, "age"] = None
print(df_clean1)
```

```python
df_clean1["age"] = df_clean1["age"].fillna(df_clean1["age"].median())
```

```python
df_clean1["score"] = df_clean1["score"].fillna(df_clean1["score"].median())
```

```python
print(df_clean1)
```
