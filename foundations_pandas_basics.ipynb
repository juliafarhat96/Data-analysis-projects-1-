```python
def average(number):
    average = sum(number) / len(number)
    return average

list1 = []
while True:
    ans1 = input("Number:")
    if ans1.lower() == "done":
        break 
    else:
        ans1 = int(ans1)
        list1.append(ans1)

result = average(list1)
print(result)
```

```python
phonebook = {}

while True:
    print("MENU")
    print("1. Add to phonebook")
    print("2. Find number")
    print("3. Quit")

    choice = input("Number of choice:")

    if choice == "1":
        choice_name = input("Name:")
        choice_number = input("Number:")
        phonebook[choice_name] = choice_number 
    elif choice == "2":
        choice_find = input("Name:")
        if choice_find in phonebook:
            print(phonebook[choice_find])
        else:
            print("Name not found")
    elif choice == "3":
        break 
    else:
        print("invalid")

```

```python
import pandas as pd 
data6 = {
    "names" : ["Amanda", "Austin" , "Shawn", "Ben", "Tessa", "Stef", "Lily", "Harry", "Mary"],
    "ages" : [19, 30, 22, 54, 33, 25, 46, 13, 15]
}
df6 = pd.DataFrame(data6)
```

```python
print(df6)
```

```python
df6["blood pressure"] = [116, 115, 130, 120, 114, 119, 145, 112, 115]
```

```python
print(df6)
```

```python
df6["Hypertension"] = df6["blood pressure"]>=130
print(df6)
```

```python
df6 = df6.drop(columns = ["Notes"])
```

```python
print(df6)
```

```python
def classify_bp(bp):
    if bp < 120:
        return "Normal"
    if bp <129:
        return "Elevated"
    if bp < 139:
        return "Hypertension stage 1"
    if bp >= 140:
        return "Hypertension stage 2"
        
```

```python
df6["BP classification"] = df6["blood pressure"].apply(classify_bp)
print(df6)
```

```python
df6 = df6.drop(columns = ["Hypertension"])
```

```python
import numpy as np 
df6.loc[2, "ages"] = np.nan 
```
